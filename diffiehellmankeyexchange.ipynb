```python
# Diffie-Hellman Key Exchange:

# we first begin with a brief description of the diffie-hellman key exchange protocol.
# two people, named Alice and Bob, need to secretly communicate some integer (called
# 'the key').
# a third party, named Eve, is the (aptly named) eavesdropper. it is assumed that any
# information shared between Alice and Bob must be sent through some open channel and
# can therefore be seen by Eve. as a result, we must be absolutely certain that at no
# point in the key exchance does Eve have sufficient information to figure out the key.


E = EllipticCurve(GF(2^221 - 3), [0, 117050, 0, 1, 0]) # this curve is called NIST M-221
# this equation defines the elliptic curve y^2 = x^3 + 117050x^2 + x over the finite
# field F_{2^221 - 3}
P = E([4, 1630203008552496124843674615123983630541969261591546559209027208557])
# in addition to the above curve, Alice and Bob also agree on a particular point P on the
# curve.
# since this information is agreed upon in advance, we assume that Eve has all this
# information. that is, she knows the elliptic curve E, the field F it is defined over,
# and the particular point P that Alice and Bob agree upon.

# Alice begins by randomly generating an element of the finite field which, for all intents
# and purposes, is an integer from 1 to 2^221 - 3
a = random.randint(1, 2^221 - 3) # only known by A
# Bob does the same
b = random.randint(1, 2^221 - 3) # only known by B
# crucially, the numbers a and b are not known by Eve.

# given the point P, Alice and Bob will each multiply P by their respective secret integers
# and send each other the result
Pa = a * P # computed by A, sent to B
Pb = b * P # computed by B, sent to A

# at this point, Eve knows the following information:
# the elliptic curve E, the finite field F, and the points P, Pa, and Pb

# then, using their secret integers, Alice and Bob will each compute the common, secret
# point Pab. thanks to the well-behavedness of point-multiplication on elliptic curves,
# we know that Pab = Pba, so Alice and Bob share the same secret.
Pba = a * Pb # computed by A
Pab = b * Pa # computed by B

# because the point Pab has never been shared through the open channel, Eve does not know
# it. moreover, it is generally believed to be true that, given E, F, P, a * P, and b * P,
# it is computationally infeasible to compute ab * P.
```
